# E4 — Robustez concentrada en Ω₁ (cinco semillas)

**Exclusivamente Ω₁, con cinco semillas**. Solo se entrenará Bi-HYCO **con interacción λ=1**, además de baselines clásicos GN-dist y GN ponderado (sin entrenamiento neuronal). Casos: (δ=3%, η=0); (δ=1%, η=0); (δ=3%, η=0.10). El primer caso reutiliza las tres semillas λ=1 de E2 y entrena únicamente dos semillas nuevas; los otros añaden cinco cada uno.

**Teoría:** sensibilidad empírica a la intensidad de la heterogeneidad y al ruido, no una cota PDE de estabilidad. A η=0 las cinco semillas miden variación del algoritmo, no cinco ruidos independientes.

**Entregables:** medias ± desviaciones, diferencias emparejadas, puntos individuales y tiempos. No se afirma robustez para Ω₂.
