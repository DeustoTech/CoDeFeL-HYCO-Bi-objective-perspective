# Allen–Cahn: complete the same experiment with five seeds

**Existing (read-only): 7, 11, 23. Additional: 37, 53.**

This notebook adds exactly two paired observation/training realizations of the
**two-parameter** Allen–Cahn problem. It preserves GN-DIST, joint-Adam Bi-HYCO,
XPINN-C, all numerical and optimization settings, and validation-based selection.
It reuses the calibrated reference, refinement check and completed three-seed
outputs. It does **not** initialize new networks from trained existing networks.

Extract the complete package in the same working folder as the original
`AllenCahn_TwoParameter_Results` directory. Keep the new `.py` and
`_five_seed_support/` beside this notebook. Do not run another notebook that writes
to these seed folders concurrently.

All code and output descriptions are in English. The full new-seed trainings
have not been run in the supplied notebook; its output cells are intentionally empty.

````python
from pathlib import Path
from IPython.display import display
from AllenCahn_Complete_FiveSeeds import FiveSeedCampaign

MODE = "pilot"  # "smoke" is a small functionality check in a SEPARATE directory
ROOT = "AllenCahn_TwoParameter_Results"
NEW_SEEDS = (37, 53)  # chosen before seeing their results; distinct from 7, 11, 23
RUN_SENSITIVITY = True  # post-training diagnostics only; never used for selection

print("Working directory:", Path.cwd())
````

## 1. Verify the three completed realizations

The next cell reads the actual saved runs, not rounded notebook output. It checks
that all three original methods finished for each existing seed, that their
configuration and datasets match the frozen experiment, and that the reference
cache is present. A source hash prevents accidentally mixing numerical versions.

No training of seeds 7, 11 or 23 is allowed in `pilot` mode. If a check fails,
fix `ROOT` or restore the original files; do not delete old results to bypass it.

For `smoke` mode only, tiny baseline runs are generated if necessary under
`ROOT + "_SMOKE_ONLY"`. They do not access or replace any pilot output.

````python
campaign = FiveSeedCampaign(root=ROOT, new_seeds=NEW_SEEDS, mode=MODE)
display(campaign.settings_table())
display(campaign.status())
print("Five paired seeds:", campaign.seeds)
print("Shared reference:", campaign.shared_reference)
print("Final report:", campaign.report)
````

## 2. Run the first additional seed (37)

The three methods run sequentially with the same new observation datasets.
Completed methods are reused. Interrupted neural methods resume their own weights,
Adam states, random-number states and selection records. An original early stop
counts as completion; this notebook does not force training past it.

All per-seed trajectories, selected and last checkpoints, histories, metrics,
600-dpi figures and sensitivity diagnostics follow the original output format.

````python
seed_37_results = campaign.run_seed(NEW_SEEDS[0], run_analysis=RUN_SENSITIVITY)
columns = ["method", "seed", "e_nu", "e_alpha_background", "e_gamma_joint",
           "physical_state_error_st", "neural_state_error_st", "time_seconds"]
display(seed_37_results[[c for c in columns if c in seed_37_results.columns]])
````

## 3. Run the second additional seed (53)

The reference equation and its discrepancy remain unchanged. Observation
locations, network initialization and resampling streams change with the seed.
The model initial parameter values and every training setting remain fixed.

````python
seed_53_results = campaign.run_seed(NEW_SEEDS[1], run_analysis=RUN_SENSITIVITY)
display(seed_53_results[[c for c in columns if c in seed_53_results.columns]])
````

## 4. Aggregate exactly five complete seeds for every method

The final report requires **15 complete method–seed combinations**, without
silently discarding incomplete or failed runs. Mean errors are averages of
per-run errors, not errors of averaged coefficient estimates. The sample standard
deviation uses `ddof=1` (denominator **4**).

Parameter estimates and Bi-HYCO-PHY use physical-validation checkpoints.
Bi-HYCO-SYN and XPINN-C states use their neural-validation checkpoints. The
main XPINN state row contains **only its direct neural prediction**. Bi-HYCO is
charged one training time, not one time for each output.

The reference was used in development and seed 7 was already a development run;
this is a descriptive five-realization robustness study, not a new blind validation.
The main three-seed report and its models remain unchanged.

````python
results = campaign.aggregate(make_summary_plots=True)
display(results["parameter_table"])
display(results["state_and_cost_table"])
display(results["metrics_per_seed"][[c for c in columns if c in results["metrics_per_seed"].columns]])
print("Report directory:", results["report_directory"])
print("Existing runs and shared reference unchanged:", campaign.verify_existing_unchanged())
````
